```python

print("Hello")

write_file = open("../data/3724-Device-Data-Fix.csv", "w+")

print("Hello")

with open("../data/3724-Device Data.csv", "r") as f:
    header = next(f)
    write_file.write(header.replace("\"", ""))
    for line in f:
        day, date, data = line.split(",", 2)
        date = day + date + "\""
        data = data.replace("\"", "")
        fixed_line = date + "," + data
        write_file.write(fixed_line)
```
